**1. Tentukan funnel yang relevan (misalnya: Visit → Add to Cart → Purchase).**

Jawaban :
Funnel yang digunakan adalah Leads → Bookings → Free Trials → Register New User. Funnel ini menggambarkan perjalanan customer dari tahap awal sebagai lead hingga melakukan registrasi sebagai pengguna baru.


```python
funnel = df.groupby("Stage")["Count"].sum()

funnel
```

```python
stage_order = [
    "Leads",
    "Bookings",
    "Free Trials",
    "Register New User"
]

funnel = funnel.reindex(stage_order)

funnel
```

```python
conversion = funnel / funnel.shift(1) * 100
conversion
```

```python
dropoff = 100 - conversion
dropoff
```

**Identifikasi drop-off terbesar dan jelaskan penyebabnya?**
 Jawabanya : Drop-off terbesar terjadi pada tahap Bookings → Free Trials dengan drop-off rate 37.32%. Data menunjukkan bahwa dari 417.400 bookings hanya 261.630 yang melanjutkan ke free trial. Dataset tidak menyediakan informasi mengenai alasan customer berhenti, sehingga penyebab spesifik tidak dapat dipastikan. Beberapa hipotesis yang perlu diuji lebih lanjut adalah proses aktivasi free trial yang kurang mudah, kurangnya follow-up setelah booking, atau ketidaksesuaian ekspektasi customer terhadap layanan.

```python
pivot = df.pivot(
    index="Customer Group",
    columns="Stage",
    values="Count"
)

pivot
```

```python
pivot["Leads_to_Bookings"] = (
    pivot["Bookings"] / pivot["Leads"] * 100
)

pivot["Bookings_to_Free_Trials"] = (
    pivot["Free Trials"] / pivot["Bookings"] * 100
)

pivot["Free_Trials_to_Register"] = (
    pivot["Register New User"] / pivot["Free Trials"] * 100
)

pivot
```

```python
funnel_result = pd.DataFrame({
    "Stage": stage_order,
    "Count": funnel.values
})

funnel_result["Conversion Rate"] = [
    None,
    funnel["Bookings"] / funnel["Leads"] * 100,
    funnel["Free Trials"] / funnel["Bookings"] * 100,
    funnel["Register New User"] / funnel["Free Trials"] * 100
]

funnel_result["Drop-off Rate"] = [
    None,
    100 - funnel_result.loc[1, "Conversion Rate"],
    100 - funnel_result.loc[2, "Conversion Rate"],
    100 - funnel_result.loc[3, "Conversion Rate"]
]

funnel_result
```

```python
overall_conversion = (
    funnel["Register New User"] /
    funnel["Leads"] * 100
)

overall_conversion
```

```python
customer_group_result = pivot[
    [
        "Leads",
        "Bookings",
        "Free Trials",
        "Register New User",
        "Leads_to_Bookings",
        "Bookings_to_Free_Trials",
        "Free_Trials_to_Register"
    ]
].copy()

customer_group_result.round(2)
```

```python
pivot["Overall_Conversion"] = (
    pivot["Register New User"] /
    pivot["Leads"] * 100
)

pivot[
    ["Leads", "Bookings", "Free Trials",
     "Register New User", "Overall_Conversion"]
].round(2)
```

```python
best_group = pivot["Overall_Conversion"].idxmax()
best_rate = pivot["Overall_Conversion"].max()

print("Customer Group dengan conversion tertinggi:")
print(best_group)
print("Overall Conversion:", round(best_rate, 2), "%")
```

```python
lowest_group = pivot["Overall_Conversion"].idxmin()
lowest_rate = pivot["Overall_Conversion"].min()

print("Customer Group dengan conversion terendah:")
print(lowest_group)
print("Overall Conversion:", round(lowest_rate, 2), "%")
```
